```python
import matplotlib.pyplot as plt
months=["January","February","March","April"]
sales=[2000,7000,3000,9000]
plt.plot(months,sales)
plt.title("sales Analysis")
plt.xlabel("months");
plt.ylabel("sales")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
data = pd.DataFrame({
    "Department": [
        "HR", "HR", "HR", "HR", "HR",
        "IT", "IT", "IT", "IT", "IT",
        "Finance", "Finance", "Finance", "Finance", "Finance",
        "Marketing", "Marketing", "Marketing", "Marketing", "Marketing"
    ],
    "Salary": [
        35000, 40000, 42000, 38000, 45000,
        55000, 60000, 65000, 58000, 70000,
        50000, 55000, 52000, 60000, 58000,
        40000, 45000, 48000, 42000, 50000
    ]
})
sns.boxplot(x="Department",y="Salary",data=data)
plt.title("Salary per department")
plt.show()
```

```python
import seaborn as sns
tips=sns.load_dataset("tips")
tips.head()
```

```python
import matplotlib.pyplot as plt
plt.hist(
    tips["total_bill"],
    bins=15,
    edgecolor="blue"
)
```

```python
import matplotlib.pyplot as plt
counts = tips["smoker"].value_counts()
plt.pie(
    counts,
    labels=counts.index,
    autopct="%1.2f%%",
    startangle=45
)
```

```python
import seaborn as sns
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex"
)
```

```python
import seaborn as sns
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="smoker"
)
```

```python
import seaborn as sns
sns.boxplot(
    data=tips,
    x="sex",
    y="total_bill"
)
```

```python
import seaborn as sns
sns.countplot(
    data=tips,
    x="day"
)
```

```python
import seaborn as sns
sns.swarmplot(
    data=tips,
    x="sex",
    y="tip"
)
```
